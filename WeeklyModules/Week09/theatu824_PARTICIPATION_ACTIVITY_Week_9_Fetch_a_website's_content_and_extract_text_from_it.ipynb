```python
#Parsing HTML
url = "https://commons.wikimedia.org/wiki/Category:Paintings_by_Edgar_Degas"

headers = {
    "User-Agent": "MSSP607 student image downloader"
}

response = requests.get(url, headers=headers)
print(response.status_code)
print(type(response))
response.raise_for_status()
soup = bs4.BeautifulSoup(response.text, "html.parser")
print(type(soup))
```

```python
# Find only the artwork images inside Wikimedia's gallery boxes.
image_tags = soup.select(".gallerybox img")
print(len(image_tags))
for image in image_tags[:3]:
    print(image.get("src"))
```

```python
image_urls = []

for image in image_tags:
    image_url = image.get("src")
    # Remove tracking parameters after the question mark.
    clean_url = image_url.split("?")[0]
    image_urls.append(clean_url)

print(len(image_urls))

for url in image_urls[:3]:
    print(url)
```

```python
#Create a folder and download the images.
import os
output_folder = "degas_paintings"
os.makedirs(output_folder, exist_ok=True)
# Select the first image URL for testing.
test_url = image_urls[0]
image_response = requests.get(test_url,headers=headers,timeout=30)
test_path = os.path.join(output_folder, "test_image.jpg")
print(test_path)
with open(test_path, "wb") as file:
    file.write(image_response.content)
print("Downloaded:", test_path)
```

```python
print(image_response.status_code)
print(image_response.headers.get("Content-Type"))
print(len(image_response.content))
```

```python
# Check the response types represented by the URLs.
extensions = set()

for url in image_urls:
    extension = url.split(".")[-1].lower()
    extensions.add(extension)

print(extensions)
```

```python
# Download the first five images as a test.
for number, url in enumerate(image_urls, start=1):
    image_response = requests.get(url,headers=headers,timeout=30)
    image_response.raise_for_status()

    # Generate a unique filename.
    extension = url.rsplit(".", 1)[-1].lower()
    filename = f"degas_{number:03d}.{extension}"
    image_path = os.path.join(output_folder, filename)

    # Save the image in binary mode.
    with open(image_path, "wb") as file:
        file.write(image_response.content)

    print("Downloaded:", filename)
```
