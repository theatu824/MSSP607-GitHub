```python
#a. We will look at stock data from Facebook, Apple, Amazon, Netflix, and Google from different files.  We will combine them into a single file and store the dataframe for further analysis.
#b. Read each file into a dataframe
df_aapl = pd.read_csv("data/Week 05 Examples/aapl.csv")
df_amzn = pd.read_csv("data/Week 05 Examples/amzn.csv")
df_fb = pd.read_csv("data/Week 05 Examples/fb.csv")
df_goog = pd.read_csv("data/Week 05 Examples/goog.csv")
df_nflx = pd.read_csv("data/Week 05 Examples/nflx.csv")
```

```python
#c. Add a column to each dataframe, called 'ticker', indicating the ticker symbol it is for (Apple's is AAPL). This symbol is how you will look up a stock.  Each file name is the ticker symbol.
df_aapl["ticker"] = "AAPL"
df_amzn["ticker"] = "AMZN"
df_fb["ticker"] = "FB"
df_goog["ticker"] = "GOOG"
df_nflx["ticker"] = "NFLX"
```

```python
#d. Append each file together into a single dataframe.
combined = pd.concat([df_aapl,df_amzn, df_fb, df_goog, df_nflx], ignore_index=True)
combined["ticker"].value_counts()

#e. Save the results in a CSV file called 'faanf.csv'.
combined.to_csv("faanf.csv",index=False)
```

```python
#'fanng.csv', use type conversion to change the 'date' column into a 'datetime' and the 'volume' column into integers.  Then sort by 'date' and 'ticker.'
df_faanf = pd.read_csv("/content/MSSP607-GitHub/faanf.csv")
df_faanf["date"] = pd.to_datetime(df_faanf["date"])
df_faanf["volume"] = df_faanf["volume"].astype("int64")
df_faanf["volume"].dtype
```

```python
df_sorted_faanf = df_faanf.sort_values(by=["date", "volume"])
df_sorted_faanf
```

```python
#g. Find the seven rows within the highest value for volume.
top7_rows = df_sorted_faanf.nlargest(n=7, columns="volume")
top7_rows
```

```python
#h. Use the melt() method to make the 'date' and 'ticker' key identifiers.  Melt the remaining columns so that there are not separated.
df_melt = df_faanf.melt(id_vars=["date","ticker" ], var_name="measurement", value_name="value")
df_melt
```
