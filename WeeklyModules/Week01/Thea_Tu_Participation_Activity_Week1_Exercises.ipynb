## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
import statistics
```

```python
mean_salaries = sum(salaries) / len(salaries)
print(mean_salaries)
statistics.mean(salaries)
```

### median

```python
sorted_salaries = sorted(salaries)
middle = len(sorted_salaries) // 2

median = (sorted_salaries[middle - 1] + sorted_salaries[middle]) / 2

print(median)
statistics.median(salaries)
```

### mode

```python
mode = max(set(salaries), key=salaries.count)
print(mode)
statistics.mode(salaries)
```

### sample variance
Remember to use Bessel's correction.

```python
variance_salaries = sum([(salary - mean_salaries) ** 2 for salary in salaries]) / (len(salaries) - 1)
print(variance_salaries)
statistics.variance(salaries)
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sample_standard_deviation = (
    sum((salary - mean_salaries) ** 2 for salary in salaries)
    / (len(salaries) - 1)
) ** 0.5

print(sample_standard_deviation)
statistics.stdev(salaries)
```

## Exercise 6: Calculating more statistics
### range

```python
salary_range = max(salaries) - min(salaries)
print(salary_range)
```

### coefficient of variation

```python
sample_standard_deviation / mean_salaries
```

### interquartile range

```python
interquartile = sorted(salaries)[len(salaries) // 4 * 3] - sorted(salaries)[len(salaries) // 4]
print(interquartile)
```

### quartile coefficent of dispersion

```python
sorted_salaries = sorted(salaries)
lower_half = sorted_salaries[:50]
upper_half = sorted_salaries[50:]

q1 = (lower_half[24] + lower_half[25]) / 2
q3 = (upper_half[24] + upper_half[25]) / 2

quartile_coefficient_of_dispersion = ((q3 - q1) / (q3 + q1))
print(quartile_coefficient_of_dispersion)
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_max_scaled = [(salary - min(salaries)) / (max(salaries) - min(salaries)) for salary in salaries]
print(min_max_scaled)
```

### standardizing

```python
standardized = [(salary - mean_salaries) / sample_standard_deviation for salary in salaries]
print(standardized)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
scaled_mean = (sum(min_max_scaled) / len(min_max_scaled))
covariance = sum((salaries[i] - mean_salaries) * (min_max_scaled[i] - scaled_mean) for i in range(len(salaries))) / (len(salaries) - 1)
print(covariance)
```

### Pearson correlation coefficient ($\rho$)

```python
scaled_mean = (sum(min_max_scaled) / len(min_max_scaled))
numerator = sum((salaries[i] - mean_salaries) * (min_max_scaled[i] - scaled_mean) for i in range(len(salaries)))
salary_squared_sum = sum((salary - mean_salaries) ** 2 for salary in salaries)
scaled_squared_sum = sum((value - scaled_mean) ** 2 for value in min_max_scaled)
pearson = numerator / (salary_squared_sum * scaled_squared_sum) ** 0.5
print(pearson)
```
