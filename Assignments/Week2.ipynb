4.The tuple() function converts a list into a tuple, while the list() function converts a tuple back into a list. A list is mutable, whereas a tuple is immutable.

```python
aa_list = [1,2,3,4,5]
aa_tuple = tuple(aa_list)
print(aa_tuple)
bb_list = list(aa_tuple)
print(bb_list)
```

```python
def comma_code(items):
    if len(items) == 0:
        return ""
    elif len(items) == 1:
        return items[0]
    elif len(items) == 2:
        return items[0] + " and " + items[1]
    else:
        beginning = ", ".join(items[:-1])
        return beginning + ", and " + items[-1]


spam = ["apples", "bananas", "tofu", "cats"]
print(comma_code(spam))
```
